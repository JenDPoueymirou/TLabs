## Fraud Detection - Preprocessing
Cleaning data and engineering features based on EDA findings

```python
# Load libraries and raw dataset
import pandas as pd
import numpy as np

df = pd.read_csv("../data/fraud_transactions.csv")
print(df.shape)
```

```python
# Drop columns not useful for modeling: names are just IDs, isFlaggedFraud is a weak naive rule
df = df.drop(columns=['nameOrig', 'nameDest', 'isFlaggedFraud'])
df.shape
```

```python
# Feature: difference between old and new origin balance (fraud tends to drain accounts)
df['balance_diff_orig'] = df['oldbalanceOrg'] - df['newbalanceOrig']
df[['oldbalanceOrg', 'newbalanceOrig', 'balance_diff_orig']].head()
```

### Engineered Feature: balance_diff_orig
- `oldbalanceOrg - newbalanceOrig` — captures how much left the origin account
- EDA showed fraud tends to drain accounts to $0, this quantifies that pattern

```python
# One-hot encode transaction type
df = pd.get_dummies(df, columns=['type'], drop_first=True)
df.columns.tolist()
```

### Encoded Type
- One-hot encoded `type` with drop_first=True (CASH_IN as baseline)
- EDA showed fraud only occurs in TRANSFER/CASH_OUT — this lets the model use that signal directly

```python
# Save cleaned dataset
df.to_csv("../data/fraud_transactions_clean.csv", index=False)
print("Saved:", df.shape)
```
